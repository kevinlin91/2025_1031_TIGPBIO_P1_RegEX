# The **compile()** function: Precompile a attern
---
## Compile a regular expression pattern into a regular expression object.
> ## using re.compile() and saving the resulting regular expression object for reuse is more efficient when the expression will be used several times in a single program.
### ```re.compile(pattern, flags=0)```

````python
pattern = r"09\d+"
regex = re.compile(pattern)
````

## What are flags?
- re.A: ASCII-only matching
- re.I: ignore case
- re.L: locale dependent (avoid unless necessary)
- re.M: multi-line (anchors match per line)
- re.S: dot matches all
- re.U: Unicode matching, default
- re.X: verbose (allows multi-line readable regex with comments)
## [Regex_Flag](https://docs.python.org/3.14/library/re.html#re.RegexFlag)

````python
print("===re.A===")
print(re.findall(r"\w+", "café", re.A)) 
print(re.findall(r"\w+", "café")) 
print("===re.I===")
print(re.findall("hello", "Hello WORLD")) 
print(re.findall("hello", "Hello WORLD", re.I)) 
print("===re.M===")
print(re.findall(r"^second", "first line\nsecond line")) 
print(re.findall(r"^second", "first line\nsecond line", re.M))
````

````python
print("===re.S===")
print(re.findall(r"H.*d", "Hello\nWorld")) 
print(re.findall(r"H.*d", "Hello\nWorld", re.S))   
print("===re.U===")
print(re.findall(r"\w+", "你好 python", re.U))
print("===re.X===")
# \d{4}-d{2}-\d{2}
pattern = r"""
    \d{4}   
    -       
    \d{2}   
    -       
    \d{2}  
"""
print(re.findall(pattern, "Date: 2025-10-31", re.X))
````

# The **search()** function: Find first match anywhere in string
---
## Scan through string looking for the *first location* where the regular expression pattern produces a match, and return a corresponding Match object.
### ```re.search(pattern, string, flags=0)```

````python
print(re.search(r"cat", "The black cat sat"))
print(re.search(r"cat", "The black cat sat").group())
print(re.search(r"dog", "The black cat sat")) 
````

# The **match()** function: Only match at the beginning
---
## If zero or more characters at the beginning of string match the regular expression pattern, return a corresponding Match object. Return None if the string does not match the pattern.
### ```re.match(pattern, string, flags=0)```

````python
print(re.match(r"The", "The black cat"))  
print(re.match(r"black", "The black cat"))
if re.match(r"black", "The black cat") is None:
    print("No match")
````

# The **fullmatch()** function: Only match at the beginning
---
## If the whole string matches the regular expression pattern, return a corresponding Match object. Return None if the string does not match the pattern.
### ```re.fullmatch(pattern, string, flags=0)```

````python
print(re.fullmatch(r"\d{4}-\d{2}-\d{2}", "2025-10-31"))  
print(re.fullmatch(r"\d{4}-\d{2}-\d{2}", "Date 2025-10-31"))
````

# The **findall()** function: Return all matches as a list of strings
---
## Return all non-overlapping matches of pattern in string, as a list of strings or tuples. The string is scanned left-to-right, and matches are returned in the order found.
### ```re.findall(pattern, string, flags=0)```

````python
print(re.findall(r"\b\w{3}\b", "cat dog elephant fox")) 
print(re.findall(r"\d{4}-\d{2}-\d{2}", "Today is 2025-10-31. Yesterday was 2025-10-30. Tomorrow will be 2025-11-01"))
````

# The **finditer()** function: Return iterator of Match objects
---
## Return an iterator yielding Match objects over all non-overlapping matches for the RE pattern in string. The string is scanned left-to-right, and matches are returned in the order found.
### ```re.finditer(pattern, string, flags=0)```

````python
for re_find in re.finditer(r"\d{4}-\d{2}-\d{2}", "Today is 2025-10-31. Yesterday was 2025-10-30. Tomorrow will be 2025-11-01"):
    print(re_find)
````

````python
for index, re_find in enumerate(re.finditer(r"\d{4}-\d{2}-\d{2}", "Today is 2025-10-31. Yesterday was 2025-10-30. Tomorrow will be 2025-11-01")):
    print(f"{index+1} found:", re_find.group())
````

# The **split()** function: Split string by pattern
---
## Split string by the occurrences of pattern. If capturing parentheses are used in pattern, then the text of all groups in the pattern are also returned as part of the resulting list. If maxsplit is nonzero, at most maxsplit splits occur, and the remainder of the string is returned as the final element of the list.
### ```re.split(pattern, string, maxsplit=0, flags=0)```

````python
print(re.split(r"[,. ]+", "apple,banana orange.grape"))
print(re.split(r"[,. ]+", "apple,banana orange.grape", maxsplit=1))
````

````python
print("apple,banana orange.grape".split(','))
````

# The **sub()** function: Search and replace
---
## Return the string obtained by replacing the **leftmost** non-overlapping occurrences of pattern in string by the replacement *repl*. If the pattern isn’t found, string is returned unchanged. *repl* can be a string or a function; if it is a string, any backslash escapes in it are processed.
### ```re.sub(pattern, repl, string, count=0, flags=0)```

````python
print(re.sub(r"\d+", "#", "ID123 and Code456"))
print(re.sub(r"\b[a-z]", lambda m: m.group().upper(), "hello world, this is python regex."))
````

# Example: Web Crawler
---
## Analysis the webpage using regular expression

````python
import requests
url = "https://idv.sinica.edu.tw/tigpbio/faculty.html"
r = requests.get(url)
html_content = r.text
print(html_content)
````

````python
match = re.search(r"<title>(.*?)</title>", html_content, re.S) #re.DOTALL
# <title>TIGP-Boiinfo.</title>
if match:
    print("Page title:", match.group(1).strip())
else:
    print("No title found")
````

````python
# clean up the html content
html = re.sub(r"<(script|style)\b.*?>.*?</\1>", "", html_content, flags=re.DOTALL)
# remove all tags
html = re.sub(r"<[^>]+>", " ", html)
# collapse whitespace
html = re.sub(r"\s+", " ", html).strip()
print(html[:2000])
name_email_re = re.compile(r"Dr\.\s*([A-Za-z.\-\s]+[A-Za-z])\s+(?:\S+\s+){0,20}([A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,})")
````

### Dr\.\s*([A-Za-z.\-\s]+[A-Za-z])\s+(?:\S+\s+){0,20}([A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,})
| Part | Meaning  | Why it's needed                                                                                                              |
| :--:| :--: | :--:|
| `Dr\.` | Matches literal text **`Dr.`** | Ensures we only extract people with the title “Dr.” |
| `\s*` | Zero or more spaces after `Dr.` | Handles both `Dr.John` and `Dr.   John`|
| `([A-Za-z.\-\s]+[A-Za-z])`| **Capture group 1 → Name**:<br>• One or more letters, dots, hyphens, or spaces<br>• But MUST end with a **letter** (`[A-Za-z]`) | Ensures we capture full names like:<br>• `James C. Liao`<br>• `Wen-Hsiung Li`<br>and not accidentally stop at a space or dot |
| `\s+` | One or more spaces | Separates the name from following text |
| `(?:\S+\s+){0,20}` | **Non-capturing group**:<br>• `\S+` = one word<br>• `\s+` = following space<br>Repeated **0 to 20 times** | Allows up to **20 extra words** between name and email) |
| `([A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,})` | **Capture group 2 → Email** | Standard email-matching pattern |


````python
output_name_email = re.findall(name_email_re, html)
print(output_name_email[:5])
print(len(output_name_email))
````
